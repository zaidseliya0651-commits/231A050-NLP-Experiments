```python
print("Name: Zaid Seliya UIN: 231A050")
documents = [
    "The cat is sleeping on the sofa.",
    "The dog is playing in the garden.",
    "She is reading a book about science.",
    "The weather is sunny and pleasant today.",
    "The cat is playing with a small ball.",
    "He enjoys reading books in his free time.",
    "The children are playing in the garden.",
    "Machine learning helps computers understand language."
]
```

```python
print("Name: Zaid Seliya UIN: 231A050")
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
```

```python
print("Name: Zaid Seliya UIN: 231A050")
for i , document in enumerate(documents):
  print(f"D{i+1}: {document}")
```

```python
print("Name: Zaid Seliya UIN: 231A050")
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(documents)
print(tfidf_matrix.shape)
```

```python
print("Name: Zaid Seliya UIN: 231A050")
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out(),
    index=["D1","D2","D3","D4","D5","D6","D7","D8"]
    )
tfidf_df.round(2)
```

```python
print("Name: Zaid Seliya UIN: 231A050")
print(tfidf_matrix)
```

```python
print("Name: Zaid Seliya UIN: 231A050")
print(vectorizer.get_feature_names_out())
```

```python
print("Name: Zaid Seliya UIN: 231A050")
similarity_matrix = cosine_similarity(tfidf_matrix)
similarity_matrix
```

```python
print("Name: Zaid Seliya UIN: 231A050")
similarity_df = pd.DataFrame(
    similarity_matrix,
    index=["D1","D2","D3","D4","D5","D6","D7","D8"],
    columns=["D1","D2","D3","D4","D5","D6","D7","D8"],
)
similarity_df.round(3)
```

```python
print("Name: Zaid Seliya UIN: 231A050")
print(similarity_matrix)
```

```python
print("Name: Zaid Seliya UIN: 231A050")
highest_similarity = -1
most_similar_pair = None

for i in range(len(documents)):
  for j in range (i + 1, len(documents)):
    similarity = similarity_matrix[i][j]
    if similarity > highest_similarity:
      highest_similarity = similarity
      most_similar_pair = (i,j)

print(
    f"Most Similar documents:"
    f"D{most_similar_pair[0] + 1} and D{most_similar_pair[1] + 1}"
)

print(f"Similarity : {highest_similarity:.3f}")
```

```python
print("Name: Zaid Seliya UIN: 231A050")
def similar(D1,D2):
  vector = vectorizer.fit_transform([D1,D2])
  similarity = cosine_similarity(vector)
  return similarity
```
